```python
from sklearn.preprocessing import StandardScaler,PowerTransformer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,RandomizedSearchCV
```

```python
num_pipeline = Pipeline(steps=[
    ("skew",PowerTransformer(method="yeo-johnson")),
    ("scaler",StandardScaler())
])

num_pipeline
```

```python
preprocessing = ColumnTransformer(transformers=[
    ("num",num_pipeline,num_cols)
])

preprocessing
```

```python
X = df.drop("Outcome",axis=1)
y = df["Outcome"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,stratify=y)
```

```python
X_train_pre = preprocessing.fit_transform(X_train)
X_test_pre = preprocessing.transform(X_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout,BatchNormalization,Input
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Input(shape=(X_train_pre.shape[1],)),
    Dense(128,activation="relu"),
    Dense(64,activation="relu"),
    Dense(32,activation="relu"),
    Dense(64,activation="relu"),
    BatchNormalization(),
    Dropout(0.1),
    Dense(32,activation="relu"),
    Dense(16,activation="relu"),
    Dense(8,activation="relu"),
    Dropout(0.1),
    Dense(16,activation="relu"),
    Dense(8,activation="relu"),
    BatchNormalization(),
    Dense(1,activation="sigmoid"),
])
```

```python
model.compile(loss="binary_crossentropy",optimizer="adam",metrics=["accuracy"])
```

```python
his = model.fit(X_train_pre,y_train,validation_data=(X_test_pre,y_test),epochs=100,batch_size=64)
```

```python
from joblib import dump

model.save("model.h5")
dump(preprocessing,"preprocessing.pkl")
dump(X.columns.tolist(),"columns.pkl")
```
